# Explaining Effusion and Pneumothorax predictions with Grad-CAM, SHAP and LIME

This notebook examines the two saved single-target DenseNet121 models for **Effusion** and **Pneumothorax**. The models are not retrained here. The aim is to look at the same predictions from three different angles: Grad-CAM for deep spatial attention, SHAP for signed pixel-level contributions relative to a background set, and LIME for a local superpixel-based approximation of the model.

To keep the comparison fair, the three methods are applied to the same automatically selected TP, FP, FN and TN test cases whenever those cases are available.


```python
import random
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    f1_score,
)

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.models import densenet121, DenseNet121_Weights
```

```python
# Reproducibility
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

## Data and trained models

We first rebuild the data split, preprocessing and DenseNet121 architecture used by the single-target experiments, then reload the saved checkpoints. The test set is kept untouched.


```python
DATA_DIR = Path("data/nih")
IMAGE_DIR = DATA_DIR / "images-small"
CHECKPOINT_DIR = Path("single_target_checkpoints")

TARGETS = ["Effusion", "Pneumothorax"]

for target in TARGETS:
    path = CHECKPOINT_DIR / f"densenet121_{target}.pth"
    print(target, "->", path, "| exists:", path.exists())
```

```python
train_df = pd.read_csv(DATA_DIR / "train-small.csv")
valid_df = pd.read_csv(DATA_DIR / "valid-small.csv")
test_df = pd.read_csv(DATA_DIR / "test.csv")

print("Initial train size:", len(train_df))
print("Initial valid size:", len(valid_df))
print("Test size:", len(test_df))
```

The original training and validation tables are merged and split again at the patient level with the same random seed. This preserves the patient-disjoint design used during model development and prevents the XAI analysis from quietly introducing a different split.


```python
def has_patient_leakage(df1, df2, patient_col="PatientId"):
    return len(
        set(df1[patient_col]).intersection(set(df2[patient_col]))
    ) > 0

valid_ratio = len(valid_df) / (len(train_df) + len(valid_df))

train_valid_df = pd.concat(
    [train_df, valid_df],
    ignore_index=True,
)

patients = train_valid_df["PatientId"].unique()

train_patients, valid_patients = train_test_split(
    patients,
    test_size=valid_ratio,
    random_state=SEED,
)

train_df = train_valid_df[
    train_valid_df["PatientId"].isin(train_patients)
].reset_index(drop=True)

valid_df = train_valid_df[
    train_valid_df["PatientId"].isin(valid_patients)
].reset_index(drop=True)

print("New train size:", len(train_df))
print("New valid size:", len(valid_df))
print("Test size:", len(test_df))

print("Train / valid leakage:", has_patient_leakage(train_df, valid_df))
print("Train / test leakage :", has_patient_leakage(train_df, test_df))
print("Valid / test leakage :", has_patient_leakage(valid_df, test_df))

assert not has_patient_leakage(train_df, valid_df)
assert not has_patient_leakage(train_df, test_df)
assert not has_patient_leakage(valid_df, test_df)
```

Images are resized to 320 × 320 and normalized sample by sample, exactly as in the single-target pipeline.


```python
class SamplewiseNormalize:
    def __call__(self, image):
        mean = image.mean()
        std = image.std()
        return (image - mean) / (std + 1e-8)


common_transform = transforms.Compose([
    transforms.Resize((320, 320)),
    transforms.ToTensor(),
    SamplewiseNormalize(),
])
```

The dataset wrapper below is only used to recover validation and test images with the same preprocessing expected by the saved models.


```python
class SingleTargetChestXrayDataset(Dataset):
    def __init__(
        self,
        df,
        image_dir,
        x_col,
        target_col,
        transform=None,
        return_metadata=False,
    ):
        self.df = df.reset_index(drop=True)
        self.image_dir = Path(image_dir)
        self.x_col = x_col
        self.target_col = target_col
        self.transform = transform
        self.return_metadata = return_metadata

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        image_path = self.image_dir / row[self.x_col]
        image = Image.open(image_path).convert("RGB")

        target = torch.tensor(
            float(row[self.target_col]),
            dtype=torch.float32,
        )

        if self.transform is not None:
            image = self.transform(image)

        if self.return_metadata:
            return image, target, str(image_path), int(idx)

        return image, target


BATCH_SIZE = 8


def make_loaders_for_target(target_name, return_metadata=False):
    valid_dataset = SingleTargetChestXrayDataset(
        valid_df,
        IMAGE_DIR,
        "Image",
        target_name,
        common_transform,
        return_metadata=return_metadata,
    )

    test_dataset = SingleTargetChestXrayDataset(
        test_df,
        IMAGE_DIR,
        "Image",
        target_name,
        common_transform,
        return_metadata=return_metadata,
    )

    valid_loader = DataLoader(
        valid_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
    )

    return valid_dataset, test_dataset, valid_loader, test_loader
```

The classifier is reconstructed as DenseNet121 with a single output logit, matching the saved single-target checkpoints.


```python
def build_single_target_model(seed=SEED):
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    model = densenet121(
        weights=DenseNet121_Weights.DEFAULT
    )

    model.classifier = nn.Linear(
        model.classifier.in_features,
        1,
    )

    return model.to(device)
```

The saved checkpoints contain both the model weights and validation metadata, including the decision threshold. They should only be loaded when the checkpoint files are trusted.


```python
def load_target_checkpoint(target_name):
    checkpoint_path = (
        CHECKPOINT_DIR / f"densenet121_{target_name}.pth"
    )

    if not checkpoint_path.exists():
        raise FileNotFoundError(
            f"Checkpoint not found: {checkpoint_path}"
        )

    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=False,
    )

    model = build_single_target_model(seed=SEED)
    model.load_state_dict(checkpoint["model_state_dict"])
    model.eval()

    return model, checkpoint


models = {}
checkpoints = {}

for target in TARGETS:
    model, checkpoint = load_target_checkpoint(target)
    models[target] = model
    checkpoints[target] = checkpoint

    print("=" * 70)
    print("Loaded:", target)
    print("Best epoch:", checkpoint.get("best_epoch"))
    print(
        "Validation threshold:",
        checkpoint.get("validation_threshold"),
    )
    print(
        "Selection metric:",
        checkpoint.get("selection_metric"),
    )
```

Before explaining individual predictions, we inspect the saved validation information and recover the threshold selected for each disease.


```python
for target in TARGETS:
    checkpoint = checkpoints[target]

    print("\n", target)
    print(
        "Validation threshold:",
        checkpoint.get("validation_threshold"),
    )
    print(
        "Validation metrics:",
        checkpoint.get("validation_metrics"),
    )
```

The helper below collects probabilities from a model without changing its state.


```python
def get_predictions(model, loader):
    model.eval()

    all_targets = []
    all_probs = []

    with torch.no_grad():
        for batch in loader:
            images = batch[0].to(device)
            targets = batch[1]

            logits = model(images).squeeze(1)
            probs = torch.sigmoid(logits)

            all_targets.append(targets.numpy())
            all_probs.append(probs.cpu().numpy())

    return (
        np.concatenate(all_targets),
        np.concatenate(all_probs),
    )
```

A short test-set check is useful here: it confirms that the reloaded models and thresholds behave as expected before any explanation is produced.


```python
def binary_metrics(y_true, y_prob, threshold=0.5):
    y_true = np.asarray(y_true).astype(int)
    y_prob = np.asarray(y_prob)

    y_pred = (y_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1],
    ).ravel()

    return {
        "ROC_AUC": roc_auc_score(y_true, y_prob),
        "PR_AUC": average_precision_score(y_true, y_prob),
        "F1": f1_score(
            y_true,
            y_pred,
            zero_division=0,
        ),
        "Sensitivity": (
            tp / (tp + fn)
            if (tp + fn) > 0
            else np.nan
        ),
        "Specificity": (
            tn / (tn + fp)
            if (tn + fp) > 0
            else np.nan
        ),
    }


test_predictions = {}
rows = []

for target in TARGETS:
    (
        valid_dataset,
        test_dataset,
        valid_loader,
        test_loader,
    ) = make_loaders_for_target(target)

    y_test, p_test = get_predictions(
        models[target],
        test_loader,
    )

    threshold = checkpoints[target][
        "validation_threshold"
    ]

    metrics = binary_metrics(
        y_test,
        p_test,
        threshold=threshold,
    )

    test_predictions[target] = {
        "y_true": y_test,
        "y_prob": p_test,
        "threshold": threshold,
    }

    rows.append({
        "Disease": target,
        "Threshold": threshold,
        **metrics,
    })

test_results = pd.DataFrame(rows)
test_results
```

## Choosing cases to explain

For each disease, test predictions are divided into true positives, false positives, false negatives and true negatives using the validation-selected threshold. Representative cases are selected once and then reused throughout the notebook. This makes differences between explanation methods easier to interpret because they are looking at the same underlying prediction.


```python
xai_candidates = {}

for target in TARGETS:
    threshold = test_predictions[target]["threshold"]
    y_true = test_predictions[target]["y_true"]
    y_prob = test_predictions[target]["y_prob"]
    y_pred = (y_prob >= threshold).astype(int)

    candidate_df = test_df.copy().reset_index(drop=True)
    candidate_df["y_true"] = y_true.astype(int)
    candidate_df["y_prob"] = y_prob
    candidate_df["y_pred"] = y_pred

    def label_case(row):
        if row["y_true"] == 1 and row["y_pred"] == 1:
            return "TP"
        if row["y_true"] == 0 and row["y_pred"] == 1:
            return "FP"
        if row["y_true"] == 1 and row["y_pred"] == 0:
            return "FN"
        return "TN"

    candidate_df["case_type"] = candidate_df.apply(
        label_case,
        axis=1,
    )

    xai_candidates[target] = candidate_df

    print("\n", "=" * 70)
    print(target)
    print(candidate_df["case_type"].value_counts())
```

Within each error category, the example is chosen from the model probabilities rather than by visual inspection.


```python
def get_representative_cases(target_name):
    df = xai_candidates[target_name]

    cases = {}

    tp = df[df["case_type"] == "TP"]
    fp = df[df["case_type"] == "FP"]
    fn = df[df["case_type"] == "FN"]
    tn = df[df["case_type"] == "TN"]

    if len(tp):
        cases["TP_high_confidence"] = (
            tp.sort_values("y_prob", ascending=False).iloc[0]
        )

    if len(fp):
        cases["FP_high_confidence"] = (
            fp.sort_values("y_prob", ascending=False).iloc[0]
        )

    if len(fn):
        cases["FN_high_probability"] = (
            fn.sort_values("y_prob", ascending=False).iloc[0]
        )

    if len(tn):
        cases["TN_low_probability"] = (
            tn.sort_values("y_prob", ascending=True).iloc[0]
        )

    return cases


for target in TARGETS:
    print("\n", target)
    cases = get_representative_cases(target)

    for case_name, row in cases.items():
        print(
            case_name,
            "| image:",
            row["Image"],
            "| true:",
            row["y_true"],
            "| prob:",
            round(float(row["y_prob"]), 4),
        )
```

## Grad-CAM

Grad-CAM asks where the convolutional network found spatial evidence for the positive disease logit. Rather than showing only the final feature map, we inspect one convolutional layer from each DenseNet block. This gives a sense of how the spatial emphasis changes from earlier to deeper representations.

All maps are computed for the **positive disease logit**, including FN and TN cases, so the direction of the explanation is consistent across examples.


```python
import torch.nn.functional as F
import matplotlib.pyplot as plt

GRADCAM_OUTPUT_DIR = Path("gradcam_outputs")
GRADCAM_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Grad-CAM output directory:", GRADCAM_OUTPUT_DIR.resolve())
```

The four selected layers are the final `conv2` operations of the four DenseNet blocks.


```python
def get_gradcam_layers(model):
    # Layers analogous to the four convolutional stages
    # used in the supplied cnn_attributions.ipynb example.
    return {
        "Dense block 1": model.features.denseblock1.denselayer6.conv2,
        "Dense block 2": model.features.denseblock2.denselayer12.conv2,
        "Dense block 3": model.features.denseblock3.denselayer24.conv2,
        "Dense block 4": model.features.denseblock4.denselayer16.conv2,
    }


for target in TARGETS:
    print("\n", target)
    for name, layer in get_gradcam_layers(models[target]).items():
        print(f"{name}: {layer}")
```

The implementation stores activations and their gradients, averages the gradients spatially to obtain channel weights, and combines the weighted activations into a normalized heatmap.


```python
def compute_gradcam_multilayer(model, input_image):
    model.eval()

    layers = get_gradcam_layers(model)

    layer_activations = {}
    layer_gradients = {}

    # Same hook logic as in the supplied cnn_attributions.ipynb.
    def forward_hook(module, inp, out):
        layer_activations[module] = out.detach()

        def backward_hook(grad):
            layer_gradients[module] = grad.detach()

        out.register_hook(backward_hook)

    handles = [
        layer.register_forward_hook(forward_hook)
        for layer in layers.values()
    ]

    try:
        model.zero_grad(set_to_none=True)

        output = model(input_image)
        score = output[0, 0]  # positive disease logit

        probability = torch.sigmoid(score).item()
        logit = score.item()

        score.backward()

        cams = {}

        for layer_name, layer in layers.items():
            acts = layer_activations[layer]       # [1, C, H, W]
            grads = layer_gradients[layer]        # [1, C, H, W]

            # Global-average-pool gradients to obtain channel weights.
            weights = grads.mean(
                dim=(2, 3),
                keepdim=True,
            )                                     # [1, C, 1, 1]

            # Weighted sum of activations.
            grad_cam = (
                weights * acts
            ).sum(
                dim=1,
                keepdim=True,
            )                                     # [1, 1, H, W]

            # ReLU: keep evidence supporting the target disease.
            grad_cam = F.relu(grad_cam)

            # Upsample to the model input size.
            grad_cam = F.interpolate(
                grad_cam,
                size=input_image.shape[-2:],
                mode="bilinear",
                align_corners=False,
            )[0, 0]

            # Normalize to [0, 1] for visualization.
            grad_cam = grad_cam.detach().cpu()
            cam_min = grad_cam.min()
            cam_max = grad_cam.max()

            if (cam_max - cam_min) > 1e-12:
                grad_cam = (
                    grad_cam - cam_min
                ) / (
                    cam_max - cam_min
                )
            else:
                grad_cam = torch.zeros_like(grad_cam)

            cams[layer_name] = grad_cam.numpy()

        return cams, logit, probability

    finally:
        # Prevent duplicated hooks when Grad-CAM is run repeatedly.
        for handle in handles:
            handle.remove()
```

For visualization, the displayed radiograph is kept separate from the normalized tensor sent to the network.


```python
def prepare_xai_image(row):
    image_path = IMAGE_DIR / row["Image"]

    image_pil = Image.open(image_path).convert("RGB")

    # Model input: same preprocessing as training.
    input_tensor = common_transform(
        image_pil
    ).unsqueeze(0).to(device)

    # Human-readable image for overlay.
    display_image = image_pil.resize((320, 320))

    return image_path, image_pil, display_image, input_tensor
```

The next helper overlays the four Grad-CAM maps beside the original image and records the prediction information used for that case.


```python
def show_gradcam_case(
    target_name,
    row,
    case_label=None,
    save=True,
):
    model = models[target_name]
    threshold = float(
        checkpoints[target_name]["validation_threshold"]
    )

    (
        image_path,
        image_pil,
        display_image,
        input_tensor,
    ) = prepare_xai_image(row)

    cams, logit, probability = compute_gradcam_multilayer(
        model,
        input_tensor,
    )

    true_label = int(row["y_true"])
    predicted_label = int(probability >= threshold)

    if case_label is None:
        case_label = row.get("case_type", "case")

    fig, axes = plt.subplots(
        1,
        5,
        figsize=(20, 4.5),
    )

    axes[0].imshow(display_image)
    axes[0].set_title(
        "Original\n"
        f"True={true_label}, Pred={predicted_label}\n"
        f"P={probability:.3f}, thr={threshold:.2f}"
    )
    axes[0].axis("off")

    for ax, (layer_name, cam) in zip(
        axes[1:],
        cams.items(),
    ):
        ax.imshow(display_image)
        ax.imshow(
            cam,
            alpha=0.5,
        )
        ax.set_title(layer_name)
        ax.axis("off")

    fig.suptitle(
        f"{target_name} | {case_label} | "
        "Grad-CAM for positive disease logit",
        fontsize=13,
    )

    plt.tight_layout()

    if save:
        safe_case = str(case_label).replace(" ", "_")
        output_path = (
            GRADCAM_OUTPUT_DIR
            / f"{target_name}_{safe_case}.png"
        )
        plt.savefig(
            output_path,
            dpi=200,
            bbox_inches="tight",
        )
        print("Saved:", output_path)

    plt.show()

    return {
        "target": target_name,
        "case": case_label,
        "image": str(image_path),
        "true_label": true_label,
        "predicted_label": predicted_label,
        "probability": probability,
        "threshold": threshold,
        "logit": logit,
        "cams": cams,
    }
```

We now run Grad-CAM on the representative cases selected earlier. If a particular TP/FP/FN/TN category is absent for one disease, that category is simply omitted.


```python
gradcam_results = {}

for target in TARGETS:
    print("\n" + "=" * 80)
    print("GRAD-CAM:", target)
    print("=" * 80)

    representative_cases = get_representative_cases(target)
    gradcam_results[target] = {}

    for case_name, row in representative_cases.items():
        result = show_gradcam_case(
            target_name=target,
            row=row,
            case_label=case_name,
            save=True,
        )

        gradcam_results[target][case_name] = result
```

For a more compact figure, the deepest DenseNet block is also shown on its own for each available case.


```python
def show_last_layer_summary(target_name, save=True):
    representative_cases = get_representative_cases(target_name)

    if len(representative_cases) == 0:
        print(f"No representative cases for {target_name}")
        return

    n_cases = len(representative_cases)

    fig, axes = plt.subplots(
        2,
        n_cases,
        figsize=(5 * n_cases, 8),
        squeeze=False,
    )

    threshold = float(
        checkpoints[target_name]["validation_threshold"]
    )

    for col_idx, (case_name, row) in enumerate(
        representative_cases.items()
    ):
        (
            image_path,
            image_pil,
            display_image,
            input_tensor,
        ) = prepare_xai_image(row)

        cams, logit, probability = compute_gradcam_multilayer(
            models[target_name],
            input_tensor,
        )

        cam = cams["Dense block 4"]

        true_label = int(row["y_true"])
        predicted_label = int(
            probability >= threshold
        )

        axes[0, col_idx].imshow(display_image)
        axes[0, col_idx].set_title(
            f"{case_name}: Original\n"
            f"True={true_label}, Pred={predicted_label}, "
            f"P={probability:.3f}"
        )
        axes[0, col_idx].axis("off")

        axes[1, col_idx].imshow(display_image)
        axes[1, col_idx].imshow(
            cam,
            alpha=0.5,
        )
        axes[1, col_idx].set_title(
            "Grad-CAM: Dense block 4"
        )
        axes[1, col_idx].axis("off")

    fig.suptitle(
        f"{target_name}: representative test cases",
        fontsize=14,
    )

    plt.tight_layout()

    if save:
        output_path = (
            GRADCAM_OUTPUT_DIR
            / f"{target_name}_last_layer_summary.png"
        )
        plt.savefig(
            output_path,
            dpi=200,
            bbox_inches="tight",
        )
        print("Saved:", output_path)

    plt.show()


for target in TARGETS:
    show_last_layer_summary(target)
```

Grad-CAM should be read as model evidence, not as a lesion annotation. A highlighted region tells us where deep convolutional features contributed to the target logit; it does not by itself show that the model localized a clinically valid abnormality. The FP and FN cases are therefore as important as the correct predictions when judging whether the model is relying on plausible image regions.


## SHAP

SHAP provides a different view of the same predictions. Here `DeepExplainer` estimates how image pixels move the single disease logit away from a reference defined by training images. Positive values support the disease output and negative values oppose it.

The background is drawn only from the training set, while the explained images are the same representative test cases used for Grad-CAM.


```python
import gc

try:
    import shap
    print("SHAP version:", shap.__version__)
except ImportError as exc:
    raise ImportError(
        "The 'shap' package is required. Install it in this environment "
        "before running the SHAP section, for example with: pip install shap"
    ) from exc

SHAP_OUTPUT_DIR = Path("shap_outputs")
SHAP_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# DeepExplainer on 320x320 DenseNet121 images is substantially heavier
# than the MNIST example in the reference notebook.
# Start small, especially on CPU; increase later if resources allow.
SHAP_BACKGROUND_SIZE = 10

print("SHAP output directory:", SHAP_OUTPUT_DIR.resolve())
print("Background size:", SHAP_BACKGROUND_SIZE)
```

DenseNet121 at 320 × 320 is much heavier than the small image models often used in SHAP examples, so the background is intentionally small by default. It can be increased when memory and runtime allow.


```python
def make_shap_background(
    target_name,
    n_background=SHAP_BACKGROUND_SIZE,
    seed=SEED,
):
    train_dataset = SingleTargetChestXrayDataset(
        train_df,
        IMAGE_DIR,
        "Image",
        target_name,
        common_transform,
    )

    rng = np.random.default_rng(seed)
    n_background = min(n_background, len(train_dataset))

    indices = rng.choice(
        len(train_dataset),
        size=n_background,
        replace=False,
    )

    background = torch.stack(
        [train_dataset[int(i)][0] for i in indices],
        dim=0,
    ).to(device)

    return background, indices


for target in TARGETS:
    bg, idx = make_shap_background(target)
    print(
        target,
        "| background:",
        tuple(bg.shape),
        "| indices:",
        idx.tolist(),
    )
    del bg
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
```

The same TP, FP, FN and TN cases are converted into a small SHAP batch. Keeping case selection fixed is more informative than comparing unrelated examples across methods.


```python
def prepare_shap_cases(target_name):
    representative_cases = get_representative_cases(target_name)

    if len(representative_cases) == 0:
        raise RuntimeError(
            f"No representative test cases found for {target_name}"
        )

    tensors = []
    display_images = []
    rows = []
    case_names = []

    for case_name, row in representative_cases.items():
        (
            image_path,
            image_pil,
            display_image,
            input_tensor,
        ) = prepare_xai_image(row)

        tensors.append(input_tensor.squeeze(0).cpu())

        # SHAP image_plot only needs an image for visualization.
        # Keep the displayed X-ray in its natural [0,1] RGB scale.
        display_np = (
            np.asarray(display_image).astype(np.float32) / 255.0
        )
        display_images.append(display_np)

        rows.append(row)
        case_names.append(case_name)

    test_samples = torch.stack(tensors, dim=0).to(device)
    display_images = np.stack(display_images, axis=0)

    return (
        test_samples,
        display_images,
        rows,
        case_names,
    )


for target in TARGETS:
    samples, display_images, rows, case_names = prepare_shap_cases(
        target
    )
    print(
        target,
        "| samples:",
        tuple(samples.shape),
        "| cases:",
        case_names,
    )
```

SHAP versions differ slightly in the array shape returned for a single-output PyTorch model. The helper below reduces those variants to a common `[N, C, H, W]` form.


```python
def normalize_single_output_shap(shap_values):
    if isinstance(shap_values, list):
        if len(shap_values) != 1:
            raise ValueError(
                "Expected one SHAP output for the single-logit model, "
                f"but received {len(shap_values)} outputs."
            )
        values = np.asarray(shap_values[0])
    else:
        values = np.asarray(shap_values)

    # Newer SHAP versions may return [N, C, H, W, 1].
    if values.ndim == 5 and values.shape[-1] == 1:
        values = values[..., 0]

    if values.ndim != 4:
        raise ValueError(
            "Expected SHAP values with shape [N,C,H,W] "
            "or [N,C,H,W,1], but got "
            f"{values.shape}"
        )

    return values
```

DeepExplainer is created separately for each disease so the large background and backward-hook objects can be released before the next model is processed.


Recent PyTorch versions can conflict with DenseNet's in-place ReLU operations when SHAP installs backward hooks. The wrapper below changes only the activation storage behavior, not the learned weights, and checks that the resulting logits still match the original model before explanations are accepted.


```python
class DenseNet121ForSHAP(nn.Module):
    def __init__(self, base_model):
        super().__init__()
        self.features = base_model.features
        self.classifier = base_model.classifier

    def forward(self, x):
        features = self.features(x)

        # Torchvision DenseNet.forward uses an in-place functional ReLU.
        # SHAP DeepExplainer backward hooks are incompatible with that
        # operation in recent PyTorch versions, so use an out-of-place ReLU.
        out = F.relu(features, inplace=False)
        out = F.adaptive_avg_pool2d(out, (1, 1))
        out = torch.flatten(out, 1)
        out = self.classifier(out)
        return out


def make_shap_compatible_model(base_model):
    # Convert all module-level in-place ReLUs inside DenseNet features.
    # This changes only how activations are stored, not the learned weights.
    for module in base_model.modules():
        if isinstance(module, nn.ReLU):
            module.inplace = False

    shap_model = DenseNet121ForSHAP(base_model)
    shap_model = shap_model.to(device)
    shap_model.eval()
    return shap_model
```

```python
def compute_shap_for_target(
    target_name,
    n_background=SHAP_BACKGROUND_SIZE,
):
    base_model = models[target_name]
    base_model.eval()

    background, background_indices = make_shap_background(
        target_name,
        n_background=n_background,
        seed=SEED,
    )

    (
        test_samples,
        display_images,
        rows,
        case_names,
    ) = prepare_shap_cases(target_name)

    # Build a SHAP-safe DenseNet forward graph.
    # No weights are changed and no retraining occurs.
    shap_model = make_shap_compatible_model(base_model)

    # Verify that replacing in-place ReLUs did not alter predictions.
    with torch.no_grad():
        # The base_model's module-level ReLUs have also been made out-of-place.
        # Its own forward still contains the final functional in-place ReLU,
        # so compare with the wrapper on a cloned input before DeepExplainer hooks.
        original_logits = base_model(test_samples.clone()).squeeze(1)
        safe_logits = shap_model(test_samples.clone()).squeeze(1)

        max_logit_difference = (
            original_logits - safe_logits
        ).abs().max().item()

    print(
        f"{target_name}: max |original logit - SHAP-safe logit| = "
        f"{max_logit_difference:.3e}"
    )

    if max_logit_difference > 1e-5:
        raise RuntimeError(
            "SHAP-safe DenseNet does not reproduce the original model "
            f"closely enough (max difference={max_logit_difference:.3e})."
        )

    print(
        f"{target_name}: building DeepExplainer with "
        f"{len(background)} background images..."
    )

    explainer = shap.DeepExplainer(
        shap_model,
        background,
    )

    try:
        raw_shap_values = explainer.shap_values(
            test_samples,
            check_additivity=False,
        )
    except TypeError:
        # Compatibility with older SHAP versions.
        raw_shap_values = explainer.shap_values(
            test_samples
        )

    shap_nchw = normalize_single_output_shap(
        raw_shap_values
    )

    with torch.no_grad():
        logits = (
            shap_model(test_samples)
            .squeeze(1)
            .detach()
            .cpu()
            .numpy()
        )

        probabilities = 1.0 / (
            1.0 + np.exp(-logits)
        )

    expected_value = np.asarray(
        explainer.expected_value
    ).reshape(-1)[0]

    result = {
        "target": target_name,
        "background_indices": background_indices,
        "test_samples": test_samples.detach().cpu(),
        "display_images": display_images,
        "rows": rows,
        "case_names": case_names,
        "shap_nchw": shap_nchw,
        "logits": logits,
        "probabilities": probabilities,
        "expected_value": float(expected_value),
        "max_logit_difference": max_logit_difference,
    }

    # Release the heavy explainer/background references.
    del explainer, background, raw_shap_values, shap_model
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result
```

SHAP is the most computationally demanding part of this notebook. Each disease is handled in turn and GPU memory is cleared between runs when CUDA is available.


A near-zero difference between the original and SHAP-safe logits is printed before `DeepExplainer` runs. That check is important because the compatibility wrapper should not alter the prediction being explained.


```python
shap_results = {}

for target in TARGETS:
    print("\n" + "=" * 80)
    print("SHAP:", target)
    print("=" * 80)

    shap_results[target] = compute_shap_for_target(
        target_name=target,
        n_background=SHAP_BACKGROUND_SIZE,
    )

    result = shap_results[target]

    print("Expected value:", result["expected_value"])
    for case_name, prob in zip(
        result["case_names"],
        result["probabilities"],
    ):
        print(
            f"{case_name}: disease probability={prob:.4f}"
        )
```

The attribution arrays are converted from channels-first to channels-last only for plotting. The model itself still receives the original PyTorch tensor format.

As a numerical sanity check, the SHAP base value plus the summed attributions is compared with the model logit for each explained case.


```python
def shap_additivity_table(target_name):
    result = shap_results[target_name]

    base_value = result["expected_value"]

    rows = []
    for i, case_name in enumerate(
        result["case_names"]
    ):
        shap_sum = float(
            np.asarray(
                result["shap_nchw"][i]
            ).sum()
        )

        reconstructed_logit = (
            base_value + shap_sum
        )
        model_logit = float(
            result["logits"][i]
        )

        rows.append({
            "Disease": target_name,
            "Case": case_name,
            "Expected_Value": base_value,
            "Sum_SHAP": shap_sum,
            "Base_plus_SHAP": reconstructed_logit,
            "Model_Logit": model_logit,
            "Absolute_Difference": abs(
                reconstructed_logit - model_logit
            ),
        })

    return pd.DataFrame(rows)


shap_additivity_results = pd.concat(
    [
        shap_additivity_table(target)
        for target in TARGETS
    ],
    ignore_index=True,
)

shap_additivity_results
```

## LIME

LIME approaches the prediction locally rather than following gradients through the network. The radiograph is divided into superpixels, many perturbed versions are created by turning superpixels on and off, and a simple weighted model is fitted around the original prediction. The highlighted regions therefore show which local image segments the surrogate found most supportive of the **positive disease class**.

The classifier passed to LIME uses the same 320 × 320 resize and sample-wise normalization as the DenseNet models. It returns two probabilities, `[P(no disease), P(disease)]`, which is the interface LIME expects for a binary classifier.


```python
import subprocess
import sys

try:
    from lime import lime_image
except ImportError:
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "lime",
    ])
    from lime import lime_image

from skimage.segmentation import mark_boundaries, slic

LIME_OUTPUT_DIR = Path("lime_outputs")
LIME_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

LIME_NUM_SAMPLES = 1000
LIME_NUM_FEATURES = 10
LIME_BATCH_SIZE = 16

print("LIME output directory:", LIME_OUTPUT_DIR.resolve())
print("Perturbations per case:", LIME_NUM_SAMPLES)
```

To make the perturbations easier to compare across cases, the segmentation is fixed to SLIC with the same settings and random seed throughout the analysis. The number of perturbations is set to 1,000 by default; reducing it speeds up experimentation, while increasing it can make the local surrogate more stable.


```python
def lime_segmentation(image):
    return slic(
        image,
        n_segments=80,
        compactness=10,
        sigma=1,
        start_label=0,
        channel_axis=-1,
    )


def make_lime_predict_fn(target_name):
    model = models[target_name]
    model.eval()

    def predict(image_batch):
        tensors = []

        for image in image_batch:
            array = np.asarray(image)

            if array.dtype != np.uint8:
                if array.max() <= 1.0:
                    array = array * 255.0
                array = np.clip(array, 0, 255).astype(np.uint8)

            pil_image = Image.fromarray(array).convert("RGB")
            tensors.append(common_transform(pil_image))

        batch_tensor = torch.stack(tensors, dim=0).to(device)

        with torch.no_grad():
            logits = model(batch_tensor).squeeze(1)
            positive_prob = torch.sigmoid(logits).cpu().numpy()

        return np.column_stack([
            1.0 - positive_prob,
            positive_prob,
        ])

    return predict


def prepare_lime_image(row):
    image_path = IMAGE_DIR / row["Image"]
    image = Image.open(image_path).convert("RGB").resize((320, 320))
    return image_path, np.asarray(image)
```

Each explanation is fitted independently around one representative test image. We keep the explanation object, the superpixel weights and the model probability so the result can be inspected numerically as well as visually.


```python
def compute_lime_for_case(target_name, row, case_name):
    image_path, image = prepare_lime_image(row)
    predict_fn = make_lime_predict_fn(target_name)

    explainer = lime_image.LimeImageExplainer(
        random_state=SEED
    )

    explanation = explainer.explain_instance(
        image.astype(np.double) / 255.0,
        classifier_fn=predict_fn,
        labels=(1,),
        hide_color=0,
        num_samples=LIME_NUM_SAMPLES,
        batch_size=LIME_BATCH_SIZE,
        segmentation_fn=lime_segmentation,
        random_seed=SEED,
    )

    probability = float(
        predict_fn(np.expand_dims(image, axis=0))[0, 1]
    )

    local_weights = explanation.local_exp.get(1, [])

    return {
        "target": target_name,
        "case": case_name,
        "image": str(image_path),
        "image_array": image,
        "probability": probability,
        "explanation": explanation,
        "local_weights": local_weights,
    }


lime_results = {}

for target in TARGETS:
    print("\n" + "=" * 80)
    print("LIME:", target)
    print("=" * 80)

    lime_results[target] = {}
    representative_cases = get_representative_cases(target)

    for case_name, row in representative_cases.items():
        result = compute_lime_for_case(
            target_name=target,
            row=row,
            case_name=case_name,
        )
        lime_results[target][case_name] = result
        print(
            f"{case_name}: disease probability="
            f"{result['probability']:.4f}"
        )
```

For the figures, only superpixels with positive support for the disease class are outlined. This keeps the visual question simple: *which local regions made the model more likely to predict the disease?* The signed segment weights are still saved separately, so negative evidence is not discarded from the numerical output.


```python
def show_lime_summary(target_name, save=True):
    results = lime_results[target_name]

    if not results:
        print(f"No representative cases for {target_name}")
        return

    n_cases = len(results)

    fig, axes = plt.subplots(
        2,
        n_cases,
        figsize=(5 * n_cases, 8),
        squeeze=False,
    )

    for col_idx, (case_name, result) in enumerate(results.items()):
        image = result["image_array"]
        explanation = result["explanation"]

        lime_image_array, mask = explanation.get_image_and_mask(
            label=1,
            positive_only=True,
            num_features=LIME_NUM_FEATURES,
            hide_rest=False,
        )

        if lime_image_array.max() > 1.0:
            lime_display = lime_image_array / 255.0
        else:
            lime_display = lime_image_array

        axes[0, col_idx].imshow(image)
        axes[0, col_idx].set_title(
            f"{case_name}: original | p={result['probability']:.3f}"
        )
        axes[0, col_idx].axis("off")

        axes[1, col_idx].imshow(
            mark_boundaries(lime_display, mask)
        )
        axes[1, col_idx].set_title(
            f"{case_name}: LIME positive regions"
        )
        axes[1, col_idx].axis("off")

    fig.suptitle(
        f"{target_name}: LIME explanations",
        fontsize=14,
    )

    plt.tight_layout()

    if save:
        output_path = (
            LIME_OUTPUT_DIR
            / f"{target_name}_LIME_representative_cases.png"
        )

        plt.savefig(
            output_path,
            dpi=200,
            bbox_inches="tight",
        )

        print("Saved:", output_path)

    plt.show()


for target in TARGETS:
    show_lime_summary(target)
```

## Reading the explanations together

The three methods are deliberately complementary. **Grad-CAM** gives a coarse view of spatial evidence in deep convolutional features. **SHAP** estimates signed contributions relative to a training-image background. **LIME** perturbs image regions and fits a simple local surrogate around one prediction. Agreement can make a qualitative pattern more convincing, but disagreement is also useful because the methods make different approximations.

None of these methods turns a model explanation into a clinical localization. The most informative comparison is therefore case-by-case: inspect TP, FP, FN and TN examples for both diseases, look for repeated reliance on plausible or implausible regions, and avoid presenting only visually attractive successes.
